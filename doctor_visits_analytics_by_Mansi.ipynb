Find out the correlation between variables in the given dataset

```python
plt.figure(figsize=(10,10))
sns.heatmap(df.corr(numeric_only=True),cbar=True,annot=True,cmap='Blues')
```

Analyse how the income of a patient affects the number of visits of the hospital

```python
plt.figure(figsize=(10,10))
sns.scatterplot(x='income',y='visits',data=df)
plt.xlabel('Income')
plt.ylabel('Visits')
plt.show()
```

Count and visualize the numbers of male and female affected by illness

```python
sns.histplot(df.gender, bins=2)
plt.show()
```

Visualize the percentage of people getting government health insurance due to low income, due to old age and also the percentage of people having private health insurance

```python
label = ['yes','no']
Y = df[df['freepoor']=='yes']
N = df[df['freepoor']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title('People getting Government Health Insurance due to low income')
plt.show()

Y = df[df['private']=='yes']
N = df[df['private']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title('Peole having Private Health Insurance')
plt.show()

Y = df[df['freepoor']=='yes']
N = df[df['freepoor']=='no']
x = [Y.shape[0],N.shape[0]]
plt.figure(figsize=(5,5))
plt.pie(x,labels=label)
plt.title('Pople getting Government health insurancr due to old age, disability or veteran status')
plt.show()
```

Plot a horizontal bar cgart to analyze the reduced days of activity due to illness based on  gender

```python
db = df.groupby('gender')['reduced'].sum().to_frame().reset_index()
plt.barh(db['gender'],db['reduced'],color = ['cornflowerblue','lightseagreen'])
plt.title('Reduced Days of Activity due to Illness')
plt.xlabel('Reduced activity')
plt.ylabel('Gender')
plt.show()
```
